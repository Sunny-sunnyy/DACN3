```python
# Cell 12: Estimate giá trị thực cho mỗi deal
# Sử dụng deal_selection từ Cell 10 (Full Pipeline Test)

from typing import List

def estimate_deals(deal_selection, ensemble) -> List[Opportunity]:
    """
    Use EnsembleAgent to estimate true value for each deal.
    Returns List[Opportunity] sorted by discount (highest first).
    """
    opportunities = []
    
    print("=" * 60)
    print("ESTIMATING TRUE VALUE FOR EACH DEAL")
    print("=" * 60)
    
    for i, deal in enumerate(deal_selection.deals, 1):
        print(f"\n[{i}/{len(deal_selection.deals)}] Estimating...")
        print(f"   📦 {deal.product_description[:60]}...")
        
        # Gọi EnsembleAgent để dự đoán giá trị thực
        estimate = ensemble.price(deal.product_description)
        
        # Tính discount
        discount = estimate - deal.price
        
        # Tạo Opportunity
        opportunity = Opportunity(
            deal=deal,
            estimate=estimate,
            discount=discount
        )
        opportunities.append(opportunity)
        
        # Hiển thị kết quả
        print(f"   💰 Sale Price: ${deal.price:.2f}")
        print(f"   📊 Estimate:   ${estimate:.2f}")
        print(f"   🏷️  Discount:   ${discount:.2f}")
    
    # Sắp xếp theo discount giảm dần
    opportunities.sort(key=lambda x: x.discount, reverse=True)
    
    return opportunities

# Chạy estimate
opportunities = estimate_deals(deal_selection, ensemble)

print(f"\n{'=' * 60}")
print(f"✅ ESTIMATED {len(opportunities)} DEALS")
print(f"{'=' * 60}")
```

```python
# Cell 13: Hiển thị bảng kết quả - Sorted by Discount

print("=" * 80)
print("🏆 BEST DEALS RANKING (Sorted by Discount - Highest First)")
print("=" * 80)

for i, opp in enumerate(opportunities, 1):
    deal = opp.deal
    discount_pct = (opp.discount / opp.estimate * 100) if opp.estimate > 0 else 0
    
    # Color code dựa trên discount
    if opp.discount > 200:
        status = "🔥 HOT DEAL!"
    elif opp.discount > 100:
        status = "✅ Good Deal"
    elif opp.discount > 0:
        status = "👍 OK"
    else:
        status = "❌ Overpriced"
    
    print(f"\n{'─' * 80}")
    print(f"📦 #{i}: {deal.product_description[:70]}...")
    print(f"   💰 Sale Price:   ${deal.price:.2f}")
    print(f"   📊 Estimate:     ${opp.estimate:.2f}")
    print(f"   🏷️  Discount:     ${opp.discount:.2f} ({discount_pct:.1f}%)")
    print(f"   📍 Status:       {status}")
    print(f"   🔗 URL: {deal.url}")

print(f"\n{'=' * 80}")
print(f"📊 Summary: {len([o for o in opportunities if o.discount > 0])} deals with positive discount")
print(f"           {len([o for o in opportunities if o.discount > 100])} deals with discount > $100")
```

```python
# Cell 14: Tạo DataFrame tổng hợp (optional - cần pandas)
import pandas as pd

data = []
for i, opp in enumerate(opportunities, 1):
    discount_pct = (opp.discount / opp.estimate * 100) if opp.estimate > 0 else 0
    data.append({
        "Rank": i,
        "Product": opp.deal.product_description[:50] + "...",
        "Sale $": f"${opp.deal.price:.2f}",
        "Estimate $": f"${opp.estimate:.2f}",
        "Discount $": f"${opp.discount:.2f}",
        "Discount %": f"{discount_pct:.1f}%",
        "Status": "🔥" if opp.discount > 200 else ("✅" if opp.discount > 100 else ("👍" if opp.discount > 0 else "❌"))
    })

df = pd.DataFrame(data)
display(df)
```
